# Task 9: Python Data Processing

**Objective:** Process structured records and generate useful summary statistics using Python.

**Tools:** Python, Jupyter Notebook

**Dataset:** Student Records

```python
students = [
    {"id": 1, "name": "Aarav", "age": 21, "marks": 85, "department": "CS"},
    {"id": 2, "name": "Priya", "age": 22, "marks": 92, "department": "IT"},
    {"id": 3, "name": "Rahul", "age": 21, "marks": 67, "department": "CS"},
    {"id": 4, "name": "Neha", "age": 23, "marks": 78, "department": "IT"},
    {"id": 5, "name": "Vikas", "age": 22, "marks": 56, "department": "CS"},
    {"id": 6, "name": "Ananya", "age": 21, "marks": 95, "department": "IT"},
    {"id": 7, "name": "Karan", "age": 22, "marks": 73, "department": "CS"},
    {"id": 8, "name": "Simran", "age": 23, "marks": 88, "department": "IT"}
]

print("Total student records:", len(students))
print(students[:2])
```

## 1. Display the Dataset

Each student is stored as a dictionary, and all dictionaries are stored in a list.

```python
for student in students:
    print(student)
```

```python
# Display the first five records

print("First five raw student records:\n")

pprint(students[:5])

print("\nTotal records:", len(students))
```

```python
def clean_records(records):
    cleaned = []

    for student in records:
        if (
            student["name"].strip()
            and 0 <= student["marks"] <= 100
            and student["age"] > 0
        ):
            cleaned.append(student)

    return cleaned


cleaned_students = clean_records(students)

print("Original records:", len(students))
print("Valid records:", len(cleaned_students))
```

## 3. Calculate Summary Statistics

Calculate average, highest, lowest, and total marks using a reusable function.

```python
def calculate_statistics(records):
    marks = [student["marks"] for student in records]

    return {
        "total_students": len(records),
        "average_marks": round(sum(marks) / len(marks), 2),
        "highest_marks": max(marks),
        "lowest_marks": min(marks),
        "total_marks": sum(marks)
    }


statistics = calculate_statistics(cleaned_students)

for key, value in statistics.items():
    print(f"{key.replace('_', ' ').title()}: {value}")
```

## 4. Process Individual Records

Add a grade and pass/fail status to each student record.

**Rule:** A score of 40 or more is a pass.

```python
def assign_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "F"


processed_students = []

for student in cleaned_students:
    record = student.copy()
    record["grade"] = assign_grade(record["marks"])
    record["status"] = "Pass" if record["marks"] >= 40 else "Fail"
    processed_students.append(record)

for student in processed_students:
    print(student)
```

## 5. Department-Wise Statistics

Calculate the number of students and average marks for each department.

```python
departments = {}

for student in processed_students:
    department = student["department"]
    departments.setdefault(department, []).append(student["marks"])

for department, marks in departments.items():
    print(f"Department: {department}")
    print("Students:", len(marks))
    print("Average marks:", round(sum(marks) / len(marks), 2))
    print()
```

## 6. Filter and Sort Records

Display students scoring 80 or above, sorted from highest to lowest marks.

```python
top_students = [
    student for student in processed_students
    if student["marks"] >= 80
]

top_students = sorted(
    top_students,
    key=lambda student: student["marks"],
    reverse=True
)

for student in top_students:
    print(
        student["name"],
        "| Marks:", student["marks"],
        "| Grade:", student["grade"]
    )
```

## 7. Final Summary

Display the most important findings from the processed dataset.

```python
passed = sum(
    1 for student in processed_students
    if student["status"] == "Pass"
)

failed = len(processed_students) - passed

print("STUDENT DATA PROCESSING REPORT")
print("-" * 35)
print("Total students:", statistics["total_students"])
print("Average marks:", statistics["average_marks"])
print("Highest marks:", statistics["highest_marks"])
print("Lowest marks:", statistics["lowest_marks"])
print("Passed students:", passed)
print("Failed students:", failed)
print("Students scoring 80 or above:", len(top_students))
```
